```python
#1.Program that find maximum among three numbers
a=int(input())
b=int(input())
c=int(input())
if(a>b and a>c):
  print(f"{a} is maximum")
elif(b>c):
  print(f"{b} is maximum")
else:
  print(f"{c}  is maximum")
```

```python
#2.Print if year is a leap year
year=int(input("Enter a year:"))
if((year%4==0 and year%100!= 0) or (year%400==0)):
  print(f"{year}is a leap year")
else:
  print(f"{year} is not a leap year")
```

```python
#3.Example for match case
day=int(input())
match day:
  case 1:
    print("Sunday")
  case 2:
    print("Monday")
  case 3:
    print("Tuesday")
  case 4:
    print("Wednesday")
  case 5:
    print("Thursday")
  case 6:
    print("Friday")
  case 7:
    print("Saturday")
```

```python
#4.Adding and Multiplying 2 complex no's
a=complex(input("Enter a real value and imaginary value:"))
b=complex(input("Enter another real value and imaginary value:"))
add=a+b
multiply=a*b
print(add)
print(multiply)
```

```python
#5.Swaping two numbers
a=int(input())
b=int(input())
print(a,b)
a,b=b,a
print(a,b)
```
