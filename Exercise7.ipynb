```python
#SET OPERATIONS
a={1,2,3,4,5}
b={4,5,6,7}
print(a|b)#union
print(a&b)#intersection
print(a-b)#difference
print(a^b)#symmetric difference
```

```python
#SET METHODS
s={1,2,3}
#Add
s.add(4)
print(s)
#update
s.update([5,6,7])
print(s)
#remove
s.remove(5)
print(s)
#discard
s.discard(4)
print(s)
#pop
s.pop()
print(s)
#clear
s.clear()
print(s)
#union
a={1,2,3}
b={3,4,5,6}
print(a.union(b))
#intersection
print(a.intersection(b))
#difference
print(a.difference(b))
#symmetric difference
print(a.symmetric_difference(b))
```
