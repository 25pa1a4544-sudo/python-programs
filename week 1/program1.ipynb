```python
#This code is an example of arithematic operators
x=int(input("Enter a value:"))
y=int(input("Enter another value:"))
print("Addition of x,y is",x+y)
print("Subtraction of x,y is",x-y)
print("Multiplication of x,y is",x*y)
print("Division of x,y",x/y)
print("Modulus of x,y",x%y)
print("Exponential of x,y",x**y)
print("Floor division of x,y",x//y)
```
