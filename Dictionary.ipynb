```python
#Dictionary functions
d={"sno":28,"name":"Asha","marks":98}
print(len(d))  #len()
print(max(d))  #max()
print(min(d))  #min()
print(sorted(d))#Sorted()
print(list(d)) #list()
print(type(d)) #dict()
print(any(d))
```

```python
#Dictionary Methods
s={"sno":25,"name":"Jessy","Age":18}
print(s.get("name"))
print(s.keys())
print(s.values())
print(s.items())
s.update({"Marks":95})
print(s)
print(s.pop("Age"))
print(s)
print(s.popitem())
print(s)
s.setdefault("Age",2)
print(s)
```

```python
#program to check if given key exists in dictionary or not
d={"a":10,"b":'mahi',"c":30,"d":'sukanya'}
key=input("Enter a key:")
if key in d:
  print("Key exists")
else:
  print("not exists")
```

```python
#program to add a new key-value pair to existing dictionary
d={"a":25,"b":'santhoshi',"c":45}
d["d"]=78  #add one item only
print(d)
d.update({"rno":34,"clg":'vit'})  #add another dictionary
print(d)
```

```python
# program to find the sum of all items in a dictionary
d={"c":56,"f":78,"g":89}
t=sum(d.values())
print(t)
```

```python
#write a program that print frequency of each element in a list in following
l=[int(i) for i in input().split()]
d={}
for i in l:
   if i in d:
    d[i]+=1
   else:
    d[i]=1
print(d)
```
