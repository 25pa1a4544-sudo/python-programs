```python
#TUPLE OPERATIONS
T1=(1,2,3,4)
T2=(5,6,7,8)
T3=T1+T2
print(T3)#Concatenation
print(T2*2)#Repetition
print(T1[3])#Forward indexing
print(T3[-3])#Backward indexing
print(T3[0:4])#Slicing
print(T3[:5])
print(T3[::-1])#Reversing tuple
print(1 in T1)
print(2 not in T2)#Membership operator
for i in T3:
  print(i,end=" ")#Iteration
```

```python
#LIST METHODS
a=(1,2,3,4)
b=(5,6,73,45)
print(a.count(2))#Count
print(a.index(3))#Index
```

```python
#BUILT-IN FUNCTIONS
a=(1,2,3,4,5)
print(len(a))#len()
print(sum(a))#sum()
print(max(a))#max()
print(min(a))#min
print(tuple(reversed(a)))#reversed()
print(tuple(sorted(a)))#sorted()
print(sum(a)/len(a))#avg()
```

```python
#USING ZIP()FUNCTION
t=("n1","n2","n3")
t1=(1,2,3)
print(tuple(zip(t,t1)))
```
