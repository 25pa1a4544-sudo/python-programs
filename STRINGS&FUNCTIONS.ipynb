```python
#Program that reverse a string using function
def reverse(s):
   return s[::-1]
s=input("Enter a string:")
print("Reverse string:",reverse(s))
```

```python
#Program that reverse
def rev_str(s):
  x=""
  for i in range(len(s)-1,-1,-1):
    x=x+s[i]
  return x
print(rev_str("hello"))
```

```python
#Palindrome
def reverse(s):
   return s[::-1]
s=input("Enter a string:")
print("Reverse string:",reverse(s))
if reverse(s)==s:
  print("Palindrome")
else:
  print("not a Palindrome")
```

```python
#Function that returns multiple values
def multiple_function(*num):
  s=sum(num)
  c=len(num)
  avg=s/c
  return s,c,avg
res_sum,res_count,res_avg=multiple_function(1,2,3,4,5,6,8)
print(f"the sum of numbers is {res_sum},the count is {res_count},the average is{res_avg:.2f}")
```
