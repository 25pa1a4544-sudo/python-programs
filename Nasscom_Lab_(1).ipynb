### Q15. Indexing and slicing to build X and y

### Q16. Vectorized operations and boolean masking
Create a 1-D NumPy array of 8 exam scores (0-100) of your choice. Without using any loop:
1. Add 5 bonus marks to every score and print the result.
2. Print a boolean array showing which scores are 50 or above.
3. Print only the scores that are 50 or above using boolean masking.

```python
# Write your code here
import numpy as np
scores = np.array([45, 67, 32, 80, 55, 49, 90, 72])
bonus_scores = scores + 5
print("Scores after bonus:")
print(bonus_scores)
mask = scores >= 50
print("\nBoolean array:")
print(mask)
print("\nScores 50 or above:")
print(scores[mask])
```
