```python
#program that copy one file content to another file
with open("f1.txt","r") as f:
  data=f.read()
  print("reading done")
with open("f2.txt","w") as f2:
  f2.write(data)
  print("file copied")
```

```python
#Print each line of a file in reverse order
with open("f1.txt","r") as f:
  data=f.readlines()
  for i in data:
    print(i[::-1])
```

```python

with open("f1.txt","r") as f:
  data=f.read()
words=data.lower().split()
words.sort()
with open("f2.txt","w" )as f:
  for word in words:
    f.write(word +"\n")
print("Words sorted and written to f2.txt" )
```

```python
with open("f1.txt","r") as f:
  data=f.read().split()
  data_lower=[i.lower() for i in data]
  sort_data=sorted(data_lower)
with open("f4.txt","w") as f2:
  for i in sort_data:
    f2.write(i+ "\n")
print("done")
```

```python
#program to compute the number of characters ,words and lines a file.
with open("f1.txt","r") as f:
  data=f.read()
characters=len(data)
words=len(data.split())
lines=len(data.splitlines())
print("Number of characters:",characters)
print("Number of words:",words)
print("Number of lines:",lines)
```
